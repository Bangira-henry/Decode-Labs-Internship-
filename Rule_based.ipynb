# Rule Based Chatbot - Bangira Henry

```
def chatbot_response(t):
 t=t.lower()
 if t in ['hi','hello']: return 'Hello! How can I help?'
 elif 'how are you' in t: return 'I am great!'
 elif t in ['bye','goodbye']: return 'Goodbye!'
 else: return 'Sorry dont understand'

print(chatbot_response('hi'))
```
